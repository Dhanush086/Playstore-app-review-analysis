## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
````

````python
# Check unique values in Type column
apps_df["Type"].unique()
````

````python
# Identify the corrupted row by filtering for Type == '0'
# '0' is not a valid value for Type (should only be 'Free' or 'Paid'),
# so this filter helps us locate the exact row causing the issue
apps_df[apps_df['Type'] == '0']
````

````python
## Remove the row where Type == '0' (the corrupted row)
apps_df = apps_df[apps_df['Type'] != '0']
# Confirm the row is removed - shape should now be one row less than original
print(apps_df.shape)
````

````python
# Remove exact duplicate rows (all columns identical)
apps_df=apps_df.drop_duplicates()
# Confirm shape after removing exact duplicates
print(apps_df.shape)
````

````python
# Remove apps that repeat by name (keep first occurrence)
apps_df=apps_df.drop_duplicates(subset='App')
print(apps_df.shape)
````

````python
# Reviews column has number-like text (e.g., "967") - convert it to actual integers
apps_df['Reviews']=apps_df['Reviews'].astype(int)
# Confirm the conversion worked - dtype should now be int, not object
print(apps_df['Reviews'].dtype)
````

````python
# Remove '+' and ',' symbol from Installs column
apps_df['Installs']=apps_df["Installs"].str.replace("+","").str.replace(",","")
apps_df['Installs']
````

````python
# Convert cleaned Installs column to integer type
apps_df["Installs"]=apps_df["Installs"].astype(int)
# Confirm conversion worked
apps_df["Installs"].dtype
````

````python
# Remove '$' symbol from Price column
apps_df["Price"] = apps_df["Price"].str.replace("$", "")
````

````python
# Convert cleaned Price column to float (keeps decimal values like 4.99)
apps_df["Price"] = apps_df["Price"].astype(float)
````

````python
# Confirm conversion worked
apps_df["Price"].dtype
````

````python
# Remove 'k' and 'm'  from Size column and convert everything into same size
def convert_size(x):
    if x.endswith("k"):
        x = x.replace("k", "")
        x = float(x)
        return x / 1000
    elif x.endswith("M"):
        x = x.replace("M", "")
        x = float(x)
        return x
    else:
        return np.nan
#Apply convert_size function to every value in the Size column
apps_df["Size"]=apps_df["Size"].apply(convert_size)
````

````python
# Confirm the conversion worked
apps_df["Size"].dtype
apps_df["Size"].head(10)
````

````python
# Convert Last Updated column from text to actual datetime type
apps_df["Last Updated"]=pd.to_datetime(apps_df["Last Updated"])
#con
````

````python
#check the conversion worked
apps_df["Last Updated"].dtype
apps_df["Last Updated"].head(10)
````

````python
# Fill missing Rating values with the median (not mean), since Rating is skewed
# by a small number of very low ratings - median is more robust to outliers
apps_df['Rating']=apps_df['Rating'].fillna(apps_df["Rating"].median())

# Confirm no missing values remain in Rating
apps_df['Rating'].isnull().sum()
````

````python
# Fill missing Size values with median, since Size is right-skewed by a few large apps
apps_df['Size']=apps_df['Size'].fillna(apps_df["Size"].median())
#confirm no missing values remain in size
apps_df['Size'].isnull().sum()
````

````python
# Remove exact duplicate rows in reviews_df
reviews_df=reviews_df.drop_duplicates()
#confirm shape after removing duplicates
reviews_df.shape
````

````python
#which category has the most apps
apps_df['Category'].value_counts()
````

````python
# Get today's date
today = pd.to_datetime('today')
print(today)
````

````python
# Create a new column: number of days since the app was last updated
apps_df['Days_Since_Update'] = (today - apps_df['Last Updated']).dt.days
# Check the result
apps_df[['App', 'Last Updated', 'Days_Since_Update']].head()
````

````python
# Merge reviews_df and apps_df on the shared 'App' column
# This combines review-level data (text, sentiment) with app-level data (category, rating, installs)
merged_df = reviews_df.merge(apps_df, on='App')

# Check the result
print(merged_df.shape)
merged_df.head()
````
